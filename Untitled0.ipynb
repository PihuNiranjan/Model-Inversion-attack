```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

print("TensorFlow:", tf.__version__)
```

```python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

x_train = x_train[..., np.newaxis]
x_test = x_test[..., np.newaxis]

print(x_train.shape)
print(y_train.shape)
```

```python
plt.figure(figsize=(10,4))

for i in range(10):
    plt.subplot(2,5,i+1)
    plt.imshow(x_train[i].squeeze(), cmap="gray")
    plt.title(str(y_train[i]))
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
def build_model():
    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(28,28,1)),

        tf.keras.layers.Conv2D(
            32, 3, activation="relu"
        ),

        tf.keras.layers.MaxPooling2D(),

        tf.keras.layers.Conv2D(
            64, 3, activation="relu"
        ),

        tf.keras.layers.MaxPooling2D(),

        tf.keras.layers.Flatten(),

        tf.keras.layers.Dense(
            128, activation="relu"
        ),

        tf.keras.layers.Dense(
            10, activation="softmax"
        )
    ])

    return model
```

```python
model = build_model()

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
history = model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.1
)
```

```python
test_loss, test_acc = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("Test accuracy:", test_acc)
```

```python
plt.figure(figsize=(8,5))

plt.plot(history.history["accuracy"], label="Training")
plt.plot(history.history["val_accuracy"], label="Validation")

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("MNIST CNN Training Accuracy")
plt.legend()
plt.grid()

plt.show()
```

#Model Inversion Attack

```python
def model_inversion_attack(
    model,
    target_class,
    max_iter=2000,
    learning_rate=0.1,
    threshold=0.99
):
    """
    Model inversion attack inspired by Fredrikson et al. (2015).

    Args:
        model: Trained TensorFlow/Keras classifier
        target_class: Class we want to reconstruct (0-9)
        max_iter: Maximum optimization iterations
        learning_rate: Step size for updating the input
        threshold: Stop when target confidence reaches this value

    Returns:
        reconstructed_image
        final_confidence
        confidence_history
    """

    # Start from a blank image
    x = tf.Variable(
        tf.zeros((1, 28, 28, 1), dtype=tf.float32)
    )

    confidence_history = []

    for iteration in range(max_iter):

        with tf.GradientTape() as tape:

            # Get model prediction
            predictions = model(x, training=False)

            # Confidence assigned to target class
            target_confidence = predictions[0, target_class]

            # Fredrikson-style objective:
            # minimize 1 - confidence(target)
            loss = 1.0 - target_confidence

        # Gradient with respect to INPUT, not model weights
        gradient = tape.gradient(loss, x)

        # Update reconstructed image
        x.assign_sub(learning_rate * gradient)

        # Keep valid image pixel range
        x.assign(tf.clip_by_value(x, 0.0, 1.0))

        confidence = float(target_confidence.numpy())
        confidence_history.append(confidence)

        # Print progress
        if iteration % 100 == 0:
            print(
                f"Iteration {iteration:4d} | "
                f"Target class: {target_class} | "
                f"Confidence: {confidence:.6f}"
            )

        # Stop if sufficiently confident
        if confidence >= threshold:
            print(
                f"\nAttack converged at iteration {iteration}"
            )
            break

    # Recalculate final confidence after final image update
    final_prediction = model(x, training=False)
    final_confidence = float(
        final_prediction[0, target_class].numpy()
    )

    return (
        x.numpy(),
        final_confidence,
        confidence_history
    )
```

### Run attack for all classes

```python
reconstructed_images = []
final_confidences = []
convergence_iterations = []

for target_class in range(10):

    print("\n" + "=" * 50)
    print(f"Attacking target class: {target_class}")
    print("=" * 50)

    reconstructed, confidence, history = model_inversion_attack(
        model=model,
        target_class=target_class,
        max_iter=2000,
        learning_rate=0.1,
        threshold=0.99
    )

    reconstructed_images.append(reconstructed)

    # Final confidence for the target class
    final_confidence = confidence

    final_confidences.append(final_confidence)

    # Number of iterations actually performed
    convergence_iterations.append(len(history))

    print(
        f"Final confidence: {final_confidence:.6f}"
    )

    print(
        f"Iterations performed: {len(history)}"
    )
```

```python
for target_class in range(10):

    print(
        f"Class {target_class}: "
        f"Confidence = {final_confidences[target_class]:.6f}, "
        f"Iterations = {convergence_iterations[target_class]}"
    )
```

### All constructec images

```python
plt.figure(figsize=(12, 5))

for i in range(10):

    plt.subplot(2, 5, i + 1)

    plt.imshow(
        reconstructed_images[i][0].squeeze(),
        cmap="gray"
    )

    plt.title(
        f"Class {i}\n"
        f"Confidence: {final_confidences[i]:.3f}"
    )

    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
predicted_classes = []

for target_class in range(10):

    prediction = model.predict(
        reconstructed_images[target_class],
        verbose=0
    )

    predicted_class = np.argmax(prediction[0])

    predicted_classes.append(predicted_class)

    print(
        f"Target = {target_class} | "
        f"Predicted = {predicted_class} | "
        f"Target confidence = "
        f"{prediction[0][target_class]:.6f}"
    )
```

### Calculate inversion success rate

```python
successful_attacks = sum(
    target == predicted
    for target, predicted
    in zip(range(10), predicted_classes)
)

attack_success_rate = (
    successful_attacks / 10
) * 100

print(
    f"Model inversion success rate: "
    f"{attack_success_rate:.2f}%"
)
```
